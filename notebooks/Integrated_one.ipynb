# **Task 5 - Predictive Statistical Modelling**

# 1. Remove irrelevant features and split the dataset


```r
# remove serial.No feature. because it is not valuble info just an identifier
model_data <- df[, -1]
```

```r


# Confirm variables
str(model_data)

# Make the result reproducible
set.seed(42)

# split the dataset
train_index <- sample(
  1:nrow(model_data),
  size = 0.80 * nrow(model_data)
)

train_data <- model_data[train_index, ]
test_data  <- model_data[-train_index, ]

cat("Training observations:", nrow(train_data), "\n")
cat("Testing observations:", nrow(test_data), "\n")
```

# 2. Build the full Multiple Linear Regression model

```r
full_model <- lm(
  Chance.of.Admit ~ GRE.Score +
                    TOEFL.Score +
                    University.Rating +
                    SOP +
                    LOR +
                    CGPA +
                    Research,
  data = train_data
)

summary(full_model)
```

# 3. Evaluate the model

```r
# Predict the testing observations
mlr_predictions <- predict(
  full_model,
  newdata = test_data
)

# Actual admission chances
actual_values <- test_data$Chance.of.Admit

# Evaluation metrics
mlr_mae <- mean(abs(actual_values - mlr_predictions))

mlr_mse <- mean((actual_values - mlr_predictions)^2)

mlr_rmse <- sqrt(mlr_mse)

mlr_r2 <- 1 -
  sum((actual_values - mlr_predictions)^2) /
  sum((actual_values - mean(actual_values))^2)

cat("Test MAE:", round(mlr_mae, 4), "\n")
cat("Test MSE:", round(mlr_mse, 4), "\n")
cat("Test RMSE:", round(mlr_rmse, 4), "\n")
cat("Test R-squared:", round(mlr_r2, 4), "\n")
```

# 4.Plot

```r
plot(
  actual_values,
  mlr_predictions,
  main = "Actual vs Predicted Admission Chance",
  xlab = "Actual Chance of Admit",
  ylab = "Predicted Chance of Admit",
  pch = 19,
  col = "steelblue"
)

abline(
  0, 1,
  col = "red",
  lwd = 2
)
```

# 7. Backward variable selection

```r
backward_model <- step(
  full_model,
  direction = "backward",
  trace = FALSE
)

summary(backward_model)
```

```r
backward_predictions <- predict(
  backward_model,
  newdata = test_data
)

backward_mae <- mean(
  abs(actual_values - backward_predictions)
)

backward_rmse <- sqrt(
  mean((actual_values - backward_predictions)^2)
)

backward_r2 <- 1 -
  sum((actual_values - backward_predictions)^2) /
  sum((actual_values - mean(actual_values))^2)

cat("Backward Model MAE:", round(backward_mae, 4), "\n")
cat("Backward Model RMSE:", round(backward_rmse, 4), "\n")
cat("Backward Model R-squared:", round(backward_r2, 4), "\n")
```

# 8. Ridge, LASSO and Elastic Net

```r
install.packages("glmnet")
library(glmnet)
```

# 8.1 Prepare the data:

```r
x_train <- model.matrix(
  Chance.of.Admit ~ .,
  data = train_data
)[, -1]

y_train <- train_data$Chance.of.Admit

x_test <- model.matrix(
  Chance.of.Admit ~ .,
  data = test_data
)[, -1]
```

# 8.2 Ridge regression

```r
set.seed(42)

ridge_cv <- cv.glmnet(
  x_train,
  y_train,
  alpha = 0
)

ridge_model <- glmnet(
  x_train,
  y_train,
  alpha = 0,
  lambda = ridge_cv$lambda.min
)

ridge_predictions <- as.numeric(
  predict(ridge_model, newx = x_test)
)
coef(ridge_model)
```

# 8.3 Lasso regression

```r
set.seed(42)

lasso_cv <- cv.glmnet(
  x_train,
  y_train,
  alpha = 1
)

lasso_model <- glmnet(
  x_train,
  y_train,
  alpha = 1,
  lambda = lasso_cv$lambda.min
)

lasso_predictions <- as.numeric(
  predict(lasso_model, newx = x_test)
)

coef(lasso_model)
```

# 8.4 Elastic Net

```r
set.seed(42)

elastic_cv <- cv.glmnet(
  x_train,
  y_train,
  alpha = 0.5
)

elastic_model <- glmnet(
  x_train,
  y_train,
  alpha = 0.5,
  lambda = elastic_cv$lambda.min
)

elastic_predictions <- as.numeric(
  predict(elastic_model, newx = x_test)
)
coef(elastic_model)
```

# 9 . Compare all models

```r
calculate_metrics <- function(actual, predicted) {

  mae <- mean(abs(actual - predicted))

  rmse <- sqrt(
    mean((actual - predicted)^2)
  )

  r2 <- 1 -
    sum((actual - predicted)^2) /
    sum((actual - mean(actual))^2)

  return(c(
    MAE = mae,
    RMSE = rmse,
    R_Squared = r2
  ))
}
```

```r
model_comparison <- rbind(
  Multiple_Linear_Regression =
    calculate_metrics(actual_values, mlr_predictions),

  Backward_Selection =
    calculate_metrics(actual_values, backward_predictions),

  Ridge =
    calculate_metrics(actual_values, ridge_predictions),

  LASSO =
    calculate_metrics(actual_values, lasso_predictions),

  Elastic_Net =
    calculate_metrics(actual_values, elastic_predictions)
)

round(model_comparison, 4)
```

# **Task 7 -Critical Evaluation of Principal Component Analysis**


PCA is evaluated to determine whether the correlated admission predictors can
be represented using fewer components. Serial Number, Chance of Admit and
Research are excluded. Serial Number is only an identifier, Chance of Admit is
the response variable, and Research is binary rather than a continuous
measurement.

# 1.1 Apply pca and standarize

```r
pca_variables <- df[, c(
  "GRE.Score",
  "TOEFL.Score",
  "University.Rating",
  "SOP",
  "LOR",
  "CGPA"
)]

str(pca_variables)
```

```r
pca_model <- prcomp(
  pca_variables,
  center = TRUE,
  scale. = TRUE
)

summary(pca_model)
```

# 1.2 View the explained variance

```r
explained_variance <- summary(pca_model)$importance

round(explained_variance, 4)
```

```r
plot(
  pca_model,
  type = "l",
  main = "Scree Plot of Admission Predictors"
)
```

```r
variance_percentage <-
  pca_model$sdev^2 / sum(pca_model$sdev^2) * 100

barplot(
  variance_percentage,
  names.arg = paste0("PC", 1:length(variance_percentage)),
  main = "Variance Explained by Principal Components",
  xlab = "Principal Component",
  ylab = "Percentage of Variance Explained",
  col = "steelblue"
)

abline(
  h = 10,
  col = "red",
  lty = 2
)
```

For this dataset, the elbow should appear around PC1–PC2 because PC1 explains most of the variance.

# **Task 8 -Critical Evaluation of Bayesian Statistical Methods**



*   Naïve Bayes is a classification method
*   For demonstration:

1 = high admission chance, at least 0.75
0 = lower admission chance, below 0.75


# 1. Create the binary target

```r
nb_train <- train_data
nb_test  <- test_data

nb_train$Admit_Binary <- factor(
  ifelse(nb_train$Chance.of.Admit >= 0.75, 1, 0),
  levels = c(0, 1)
)

nb_test$Admit_Binary <- factor(
  ifelse(nb_test$Chance.of.Admit >= 0.75, 1, 0),
  levels = c(0, 1)
)

table(nb_train$Admit_Binary)
table(nb_test$Admit_Binary)
```

```r
dim(nb_test)
```

```r
nrow(nb_test)   # Number of observations
ncol(nb_test)   # Number of variables
```

```r
str(nb_test)
head(nb_test)
```

# 2. Train Naïve Bayes

```r
install.packages("e1071")
```

Research was stored as a numeric variable, Naïve Bayes treated it like a continuous normally distributed variable.So we need to convert it to factor before train the model.

```r
nb_train$Research <- factor(
  nb_train$Research,
  levels = c(0, 1),
  labels = c("No", "Yes")
)

nb_test$Research <- factor(
  nb_test$Research,
  levels = c(0, 1),
  labels = c("No", "Yes")
)
```

```r
library(e1071)

nb_model <- naiveBayes(
  Admit_Binary ~ GRE.Score +
                 TOEFL.Score +
                 University.Rating +
                 SOP +
                 LOR +
                 CGPA +
                 Research,
  data = nb_train
)

nb_model
```

# 3. Generate predictions

```r
nb_predictions <- predict(
  nb_model,
  newdata = nb_test,
  type = "class"
)

nb_probabilities <- predict(
  nb_model,
  newdata = nb_test,
  type = "raw"
)

head(nb_probabilities)
```

# 4. confusion matrix

```r
confusion_matrix <- table(
  Actual = nb_test$Admit_Binary,
  Predicted = nb_predictions
)

confusion_matrix
```

```r
TN <- confusion_matrix["0", "0"]
FP <- confusion_matrix["0", "1"]
FN <- confusion_matrix["1", "0"]
TP <- confusion_matrix["1", "1"]
```

# 5. Calculate classification metrics

```r
nb_accuracy <- (TP + TN) / sum(confusion_matrix)

nb_precision <- TP / (TP + FP)

nb_recall <- TP / (TP + FN)

nb_specificity <- TN / (TN + FP)

nb_f1 <- 2 * (
  nb_precision * nb_recall /
  (nb_precision + nb_recall)
)

cat("Accuracy:", round(nb_accuracy, 4), "\n")
cat("Precision:", round(nb_precision, 4), "\n")
cat("Recall:", round(nb_recall, 4), "\n")
cat("Specificity:", round(nb_specificity, 4), "\n")
cat("F1-score:", round(nb_f1, 4), "\n")
```
